# 1. gensim

Импортируем `gensim`

```python
from gensim.models import Word2Vec
import gensim.downloader as api
```

```python
info = api.info()
```

Проверяем список доступных моделей

```python
list(info['models'].keys())
```

Скачиваем модель word2vec

```python
model = api.load('word2vec-google-news-300')
```

Смотрим вектор слова

```python
model['mouse'][:20]
```

Измеряем схожесть слов

```python
model.similarity('car', 'cat')
```

```python
model.similarity('cat', 'kitten')
```

Находим наиболее похожие по смыслу слова

```python
model.most_similar('cat')
```

The "king - man + woman" example

```python
model.most_similar(positive=['Paris', 'Estonia'], negative=['France'])
```
